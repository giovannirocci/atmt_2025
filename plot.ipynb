```python
import matplotlib.pyplot as plt
```

```python
bleu_scores = {
    'beam_3':29.58,
    'beam_5':32.73,
    'beam_7':34.03,
    'beam_10':28.87,
    'greedy':19.49,
}

times = {
    'beam_3':35.327,
    'beam_5':37.021,
    'beam_7':42.923,
    'beam_10':54.290,
    'greedy':47.587,
}
```

```python
# Plotting BLEU scores bar chart horizontal
plt.figure(figsize=(10, 6))
plt.barh(list(bleu_scores.keys()), list(bleu_scores.values()), color='blue')
plt.xlabel('BLEU Score')
plt.title('BLEU Scores for Different Decoding Strategies')
plt.grid(axis='x')
plt.show()
```

```python
# Plotting inference times bar chart horizontal
plt.figure(figsize=(10, 6))
plt.barh(list(times.keys()), list(times.values()), color='green')
plt.xlabel('Inference Time (seconds)')
plt.title('Inference Times for Different Decoding Strategies')
plt.grid(axis='x')
plt.show()
```
